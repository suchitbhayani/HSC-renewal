```python
df15_r = pd.read_csv('correlations/GEP_15_analyzed.csv').drop('Unnamed: 0', axis=1).set_index('gene')
df22_r = pd.read_csv('correlations/GEP_22_analyzed.csv').drop('Unnamed: 0', axis=1).set_index('gene')
```

```python
hsc_15_lr = LinearRegression().fit(df15_r[['r hsc']], df15_r['GEP 15 (DNA Replication)'])
prog_15_lr = LinearRegression().fit(df15_r[['r prog']], df15_r['GEP 15 (DNA Replication)'])
hsc_22_lr = LinearRegression().fit(df22_r[['r hsc']], df22_r['GEP 22 (Cell cycle)'])
prog_22_lr = LinearRegression().fit(df22_r[['r prog']], df22_r['GEP 22 (Cell cycle)'])
```

```python
def pred_GEP_expression(df, model, cell_type, GEP):
    assert cell_type in ['hsc', 'prog']
    assert GEP in [15, 22]
    
    return df.assign(**{f'GEP {GEP} pred based on r {cell_type}': model.predict(df[[f'r {cell_type}']])})
```

```python
df15 = (df15_r
        .pipe(pred_GEP_expression, hsc_15_lr, 'hsc', 15)
        .pipe(pred_GEP_expression, prog_15_lr, 'prog', 15)
       )
df22 = (df22_r
        .pipe(pred_GEP_expression, hsc_22_lr, 'hsc', 22)
        .pipe(pred_GEP_expression, prog_22_lr, 'prog', 22)
       )
```
